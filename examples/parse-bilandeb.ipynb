```python
import martpy.outputs.bilandeb as parsedeb
```

# in a Python script

```python
help(parsedeb.extract_layer_transfers)
```

## to get a `xarray.Dataset`

```python
ds = parsedeb.extract_layer_transfers(
    marthe_file="bilandeb.txt"
)
```

```python
print(ds)
```

## to get a netCDF file

```python
parsedeb.extract_layer_transfers(
    marthe_file="bilandeb.txt",
    netcdf_file="layer-transfers.nc"
)
```

```python
import xarray as xr
ds = xr.load_dataset("layer-transfers.nc")
```

```python
print(ds)
```
